```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from mpl_toolkits.mplot3d import Axes3D
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset
df = pd.read_csv('Cleaned/finaldataset.csv')
df.head()
```

```python
# Drop non-numeric columns (like date)
df = df.drop(columns=['date'])
```

```python
# Separate features and target
target_variable = 'FEDRates'
X = df.drop(columns=[target_variable])
y = df[target_variable]
```

```python
# Standardize the features (PCA works best with standardized data)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply PCA
pca = PCA(n_components=X.shape[1])  
X_pca = pca.fit_transform(X_scaled)
```

```python
# Explained variance ratio plot
plt.figure(figsize=(10, 6))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o', linestyle='--')
plt.title('Cumulative Explained Variance by PCA Components')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.grid(True)
plt.show()
plt.savefig("PCA_Explained_Variance.png")
```

```python
explained_variance_df = pd.DataFrame({
    'Principal Component': [f'PC{i+1}' for i in range(len(pca.explained_variance_ratio_))],
    'Explained Variance Ratio': pca.explained_variance_ratio_
})
print("Explained Variance by Each Component:")
print(explained_variance_df)
```

```python
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)
```

```python
# Plot the first two components (color by FEDRates)
plt.figure(figsize=(8, 6))
plt.scatter(X_pca_2d[:, 0], X_pca_2d[:, 1], c=y, cmap='coolwarm', edgecolor='k', s=50)
plt.title('PCA Projection (First 2 Components)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(label='FEDRates')
plt.grid(True)
plt.show()
plt.savefig("PCA_2D_Projection.png")
```

```python
# For visualization, we'll reduce to first 3 components
pca_3d = PCA(n_components=3)
X_pca_3d = pca_3d.fit_transform(X_scaled)
```

```python
# Animation function for rotating 3D plot
def animate_3d(i, scatter, ax):
    ax.view_init(elev=20, azim=i)
    return scatter,
```

```python
# Plot Original Data in 3D
fig1 = plt.figure(figsize=(7, 7))
ax1 = fig1.add_subplot(111, projection='3d')
scatter1 = ax1.scatter(X.iloc[:, 0], X.iloc[:, 1], X.iloc[:, 2], c=y, cmap='coolwarm', s=50)
ax1.set_title('Original Data - First 3 Features')
ax1.set_xlabel(X.columns[0])
ax1.set_ylabel(X.columns[1])
ax1.set_zlabel(X.columns[2])
ani1 = animation.FuncAnimation(fig1, animate_3d, frames=360, interval=30, fargs=(scatter1, ax1))
```

```python
# --- Plot PCA-Transformed Data in 3D ---
fig2 = plt.figure(figsize=(7, 7))
ax2 = fig2.add_subplot(111, projection='3d')
scatter2 = ax2.scatter(X_pca_3d[:, 0], X_pca_3d[:, 1], X_pca_3d[:, 2], c=y, cmap='coolwarm', s=50)

ax2.set_title('After PCA - First 3 Components')
ax2.set_xlabel('Principal Component 1')
ax2.set_ylabel('Principal Component 2')
ax2.set_zlabel('Principal Component 3')

ani2 = animation.FuncAnimation(fig2, animate_3d, frames=360, interval=30, fargs=(scatter2, ax2))

# Show both animations side by side
plt.show()

```

```python
# Explained variance ratio in percentage
explained_variance_percentage = pca.explained_variance_ratio_ * 100

# Scree Plot in Percentage with values on top of each bar
plt.figure(figsize=(10, 6))
bars = plt.bar(range(1, len(explained_variance_percentage) + 1), explained_variance_percentage, 
               color='teal', alpha=0.7, edgecolor='k')

plt.xlabel('Principal Component')
plt.ylabel('Variance Explained (%)')
plt.title('Scree Plot (Explained Variance by PCA Components)')
plt.xticks(range(1, len(explained_variance_percentage) + 1))
plt.grid(True)

# Add percentage text on top of each bar
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 1, f'{height:.2f}%', ha='center', fontsize=10)

plt.show()
plt.savefig("PCA_Scree_Plot.png")
```

```python
# ---- Loading Plot (Feature Contributions to Components) ----
loadings = pd.DataFrame(pca.components_.T, columns=[f'PC{i+1}' for i in range(X.shape[1])], index=X.columns)

plt.figure(figsize=(12, 6))
for i in range(7):  # Show loadings for first 3 PCs
    plt.bar(loadings.index, loadings.iloc[:, i], label=f'PC{i+1}')

plt.axhline(0, color='black', lw=0.2)
plt.xticks(rotation=90)
plt.legend()
plt.title('Loading Plot - Feature Contributions to First 3 PCs')
plt.ylabel('Loading Value')
plt.grid(True)
plt.show()
plt.savefig("PCA_Loading_Plot.png")
```
